## 🌐 28 Nations Dataset Inspection & Cloud Detection Demo

```python
import os
import sys
import json
import pandas as pd

# Add Version6 scripts to sys.path
sys.path.append(os.path.abspath("./Version6/scripts"))
from cloud_detector_predictor import MultiNationCloudDetectorPredictor

print("Initializing Version 6 Multi-Nation Cloud Engine...")
predictor = MultiNationCloudDetectorPredictor(data_dir="./Version6/data/nations")

# 1. Load manifest of 28 popular nations
manifest_path = "./Version6/data/nations/nations_manifest.json"
with open(manifest_path, "r", encoding="utf-8") as f:
    nations = json.load(f)

print(f"✅ Successfully loaded manifest for {len(nations)} popular nations!")
df_nations = pd.DataFrame(nations)[["code", "name", "climate_zone", "record_count", "avg_cloud_cover_pct", "latest_cloud_type", "csv_file", "json_file"]]
display(df_nations.head(10))
```

### Execute Time Series Prediction for South Korea (KR), United States (US), and Japan (JP)

```python
sample_nations = ["KR", "US", "JP", "DE", "GB"]

for code in sample_nations:
    records = predictor.load_nation_data(code, format_type="json")
    detection = predictor.detect_cloud_formation(records[-1])
    forecast = predictor.predict_time_series(records, forecast_horizon_hours=24)
    
    print(f"=== {detection['country_name']} ({code}) ===")
    print(f"  Detected Cloud Form: {detection['detected_cloud_form']} (Confidence: {detection['confidence']*100:.1f}%)")
    print(f"  Cloud Cover: {detection['cloud_cover_pct']}% | Reflectance: {detection['satellite_reflectance']}")
    print(f"  24h Forecast Form: {forecast['predictions'][-1]['predicted_cloud_form']} | Predicted Cover: {forecast['predictions'][-1]['predicted_cloud_cover_pct']}%")
    print("-" * 50)
```
